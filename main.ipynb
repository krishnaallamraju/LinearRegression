# House Price Prediction with Linear Regression

Train a one feature linear regression model that predicts house price from area. The notebook uses a local CSV, a reproducible train/test split, and gradient descent on standardized values.


```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
# Find the dataset beside this notebook so the project works from any folder.
data_path = Path("House Price Prediction Dataset.csv")
if not data_path.exists():
    raise FileNotFoundError(
        f"Could not find {data_path.resolve()}. Place the CSV beside this notebook."
    )

data = pd.read_csv(data_path)
required_columns = {"Area", "Price"}
missing_columns = required_columns.difference(data.columns)
if missing_columns:
    raise ValueError(f"Dataset is missing required columns: {sorted(missing_columns)}")

# Keep the feature and target numeric, and remove incomplete rows.
model_data = data[["Area", "Price"]].apply(pd.to_numeric, errors="coerce").dropna()
if len(model_data) < 2:
    raise ValueError("At least two rows with numeric Area and Price are required.")

model_data.head()
```

## Split and standardize the data

Standardization keeps gradient descent stable. We calculate the means and standard deviations from the training set only, then apply them to both sets.


```python
X = model_data["Area"].to_numpy(dtype=float)
y = model_data["Price"].to_numpy(dtype=float)

rng = np.random.default_rng(42)
indices = rng.permutation(len(X))
split_at = int(len(X) * 0.8)
train_idx, test_idx = indices[:split_at], indices[split_at:]
if len(train_idx) == 0 or len(test_idx) == 0:
    raise ValueError("The dataset needs enough rows to create train and test sets.")

x_mean, x_std = X[train_idx].mean(), X[train_idx].std()
y_mean, y_std = y[train_idx].mean(), y[train_idx].std()
if x_std == 0 or y_std == 0:
    raise ValueError("Area and Price must vary to train a regression model.")

x_train = (X[train_idx] - x_mean) / x_std
y_train = (y[train_idx] - y_mean) / y_std
x_test = (X[test_idx] - x_mean) / x_std
y_test = y[test_idx]
```

## Train with gradient descent

The model is `prediction = weight * area + bias`. Each step moves the parameters in the direction that reduces mean squared error.


```python
weight = 0.0
bias = 0.0
learning_rate = 0.05
epochs = 3000
n = len(x_train)

for epoch in range(epochs):
    y_pred = weight * x_train + bias
    errors = y_pred - y_train
    d_weight = (2 / n) * np.dot(errors, x_train)
    d_bias = (2 / n) * errors.sum()
    weight -= learning_rate * d_weight
    bias -= learning_rate * d_bias

# Convert parameters back to the original Area and Price units.
price_per_area = weight * y_std / x_std
price_intercept = y_mean + y_std * bias - price_per_area * x_mean
print(f"Price ≈ {price_per_area:,.2f} × Area + {price_intercept:,.2f}")
```

```python
# Evaluate on the held-out test set.
test_pred = (weight * x_test + bias) * y_std + y_mean
mae = np.mean(np.abs(y_test - test_pred))
rmse = np.sqrt(np.mean((y_test - test_pred) ** 2))
ss_total = np.sum((y_test - y_test.mean()) ** 2)
r2 = 1 - np.sum((y_test - test_pred) ** 2) / ss_total if ss_total else float("nan")

print(f"Test MAE:  {mae:,.2f}")
print(f"Test RMSE: {rmse:,.2f}")
print(f"Test R²:   {r2:.3f}")
```

```python
# Compare a few predictions with the actual test values.
results = pd.DataFrame({
    "Area": X[test_idx],
    "Actual Price": y_test,
    "Predicted Price": test_pred,
})
results.head(10).round(2)
```
