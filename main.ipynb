```python
import pandas as pd
import numpy as np
data=pd.read_csv(r"C:\Users\sowmy\OneDrive\Documents\Ai_basic\House Price Prediction Dataset.csv")
data
```

```python
weight = 0
bias = 0
y_data = data['Price']
x_data = data['Area']

```

```python
def predit(y_data,input) :
    y_pred = weight*input+bias
    return y_pred
```

```python
def error(y_data,y_pred) :
    error = y_pred - y_data
    return error

def loss(error) :
    loss = np.sum(error**2)
    return loss
```

```python
def optim(x_data, y_data) :
    epoch = range(1000)
    global weight, bias
    learning_rate = 0.0001
    n=len(x_data)
    
    for epoch in epochs :
        pred= predit(y_data,x_data)
        errors = error(pred,y_data)
        d_weight=2/n * np.sum(errors* x_data)
        d_bias=2/n*np.sum(errors)
        weight=weight-learning_rate*d_weight
        bias=bias-learning_rate*d_bias
        losses=loss(errors)
        print(f"loss:{losses},epoch: {epoch}")
```

```python
def optim(x_data, y_data):
    global weight, bias
    learning_rate = 1e-8
    n = len(x_data)

    for epoch in range(1000):
        pred = predit(y_data, x_data)
        errors = error(y_data, pred)
        d_weight = (2 / n) * np.sum(errors * x_data)
        d_bias = (2 / n) * np.sum(errors)
        weight -= learning_rate * d_weight
        bias -= learning_rate * d_bias

        if epoch % 100 == 0:
            print(f"loss: {loss(errors)}, epoch: {epoch}")
```
